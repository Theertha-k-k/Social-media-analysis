```python
# Cleaning missing data

# Detecting missing values

df.isnull().sum()
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
# Handling missing values

# for age

df["age"] = df["age"].fillna(
    df["age"].median()
)
```

```python
# for likes,comments,share

df["likes"] = df["likes"].fillna(
    df["likes"].median()
)

df["comments"] = df["comments"].fillna(
    df["comments"].median()
)

df["shares"] = df["shares"].fillna(
    df["shares"].median()
)
```

```python
# for gender

df["gender"] = df["gender"].fillna(
    df["gender"].mode()[0]
)
```

```python
# for sentiment

df["sentiment"] = df["sentiment"].fillna(
    df["sentiment"].mode()[0]
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv("social_media_cleaned.csv")

# Convert date column
df["posted_at"] = pd.to_datetime(
    df["posted_at"],
    errors="coerce",
    dayfirst=True
)

# Create engagement score
df["engagement_score"] = (
    df["likes"] +
    df["comments"] +
    df["shares"]
)

print("Dataset Shape:", df.shape)
print(df.columns.tolist())
```

```python
# C. BEHAVIOURAL INSIGHTS

# 1. What is the best time of day for impressions
# Goal - Identify when posts receive the highest impressions

# Extract posting hour

df["posting_hour"] = df["posted_at"].dt.hour
```

```python
# Create time categories

def get_time_period(hour):

    if 5 <= hour < 12:
        return "Morning"

    elif 12 <= hour < 17:
        return "Afternoon"

    elif 17 <= hour < 21:
        return "Evening"

    else:
        return "Night"

df["time_of_day"] = df["posting_hour"].apply(
    lambda x: get_time_period(x) if pd.notna(x) else np.nan
)
```

```python
time_analysis = df.groupby("time_of_day").agg(
    total_posts=("post_id", "count"),
    avg_impressions=("impression_count", "mean"),
    total_impressions=("impression_count", "sum"),
    avg_engagement=("engagement_score", "mean")
).sort_values(
    "avg_impressions",
    ascending=False
)

time_analysis
```

```python
# Iddentify the best time

time_analysis.head(1)
```

```python
# Device type impact on watch time
# Goal- Compare watch time across different devices

# Group by device type

device_analysis = df.groupby("device_type").agg(
    total_posts=("post_id", "count"),
    avg_watch_time=("watch_time_sec", "mean"),
    avg_engagement=("engagement_score", "mean"),
    avg_impressions=("impression_count", "mean")
).sort_values(
    "avg_watch_time",
    ascending=False
)

device_analysis
```
